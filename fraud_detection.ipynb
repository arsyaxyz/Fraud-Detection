```python
# ==========================================
# PROJECT 1: FRAUD DETECTION
# STEP 1 - LOAD DATASET
# ==========================================

from google.colab import drive
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Hubungkan Google Drive
drive.mount('/content/drive')

# 2. Tentukan lokasi dataset
drive_path = "/content/drive/MyDrive/dataset/dataset cc"
print("Daftar file:", os.listdir(drive_path))

# 3. Baca dataset
file_path = os.path.join(drive_path, "creditcard.csv")
df = pd.read_csv(file_path)

# 4. Periksa dataset
print("\nUkuran dataset:", df.shape)
print("\nInformasi dataset:")
df.info()

print("\nLima data pertama:")
display(df.head())

print("\nStatistik deskriptif:")
display(df.describe().T)

# 5. Periksa missing values
print("\nMissing values:")
print(df.isnull().sum().sum())

# 6. Analisis distribusi kelas
print("\nJumlah transaksi per kelas:")
print(df["Class"].value_counts())

print("\nPersentase transaksi (%):")
print((df["Class"].value_counts(normalize=True) * 100).round(4))

# 7. Visualisasi distribusi fraud
sns.countplot(data=df, x="Class")
plt.title("Fraud vs Legitimate Transactions")
plt.xlabel("Class (0 = Normal, 1 = Fraud)")
plt.ylabel("Number of Transactions")
plt.show()
```

```python
# ==========================================
# PROJECT 1: FRAUD DETECTION
# STEP 2 - EXPLORATORY DATA ANALYSIS
# ==========================================

import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. Ringkasan dataset
total = len(df)
fraud = int((df["Class"] == 1).sum())
normal = int((df["Class"] == 0).sum())

print("=== DATASET SUMMARY ===")
print(f"Total transaksi : {total:,}")
print(f"Transaksi normal: {normal:,} ({normal / total * 100:.4f}%)")
print(f"Transaksi fraud : {fraud:,} ({fraud / total * 100:.4f}%)")
print(f"Rasio normal : fraud = {normal / fraud:.1f} : 1")
print(f"Duplikasi data : {df.duplicated().sum():,}")

# 2. Visualisasi jumlah transaksi
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.countplot(data=df, x="Class", ax=axes[0])
axes[0].set_title("Transaction Count by Class")
axes[0].set_xticks([0, 1], ["Normal", "Fraud"])
axes[0].set_xlabel("Transaction Type")
axes[0].set_ylabel("Number of Transactions")

class_pct = df["Class"].value_counts(normalize=True).sort_index() * 100
axes[1].bar(["Normal", "Fraud"], class_pct.values)
axes[1].set_title("Transaction Percentage by Class")
axes[1].set_ylabel("Percentage (%)")

plt.tight_layout()
plt.show()

# 3. Distribusi nominal transaksi
df["Amount_log"] = np.log1p(df["Amount"])

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Class", y="Amount_log", showfliers=False)
plt.title("Transaction Amount Distribution by Class")
plt.xlabel("Class (0 = Normal, 1 = Fraud)")
plt.ylabel("Log(1 + Amount)")
plt.show()

# 4. Fitur dengan korelasi terkuat terhadap Class
correlation = (
    df.drop(columns=["Amount_log"], errors="ignore")
      .corr(numeric_only=True)["Class"]
      .drop("Class")
)

top_corr = correlation.loc[
    correlation.abs().sort_values(ascending=False).head(10).index
].sort_values()

plt.figure(figsize=(9, 6))
top_corr.plot(kind="barh")
plt.title("Top 10 Features Correlated with Fraud Label")
plt.xlabel("Correlation with Class")
plt.ylabel("Feature")
plt.tight_layout()
plt.show()

print("\n=== TOP CORRELATED FEATURES ===")
print(top_corr.sort_values(key=abs, ascending=False))

# 5. Business insight awal
print("\n=== INITIAL BUSINESS INSIGHTS ===")
print(f"Fraud rate: {fraud / total * 100:.4f}%")
print("Accuracy saja tidak cukup untuk mengevaluasi model.")
print("Fokus berikutnya: Recall, Precision, F1-score, dan PR-AUC.")
```

```python
# ==========================================
# PROJECT 1: FRAUD DETECTION
# STEP 3 - DATA CLEANING & DATA SPLITTING
# ==========================================

from sklearn.model_selection import train_test_split

# 1. Salin dataset dan hapus fitur sementara dari EDA
df_clean = df.drop(
    columns=["Amount_log"], errors="ignore"
).copy()

# 2. Periksa duplikat sebelum menghapusnya
duplicate_mask = df_clean.duplicated()

print("=== DUPLICATE ANALYSIS ===")
print("Jumlah baris duplikat:", duplicate_mask.sum())

print("\nDuplikat berdasarkan kelas:")
print(
    df_clean.loc[duplicate_mask, "Class"]
    .value_counts()
    .reindex([0, 1], fill_value=0)
    .rename(index={0: "Normal", 1: "Fraud"})
)

# 3. Hapus baris yang identik sepenuhnya
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("\n=== CLEANED DATASET ===")
print("Ukuran dataset setelah cleaning:", df_clean.shape)
print("\nDistribusi kelas setelah cleaning:")
print(df_clean["Class"].value_counts())

# 4. Pisahkan fitur dan target
X = df_clean.drop(columns=["Class"])
y = df_clean["Class"]

# 5. Bagi data: 80% training dan 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# 6. Periksa hasil pembagian
print("\n=== TRAIN-TEST SPLIT ===")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("\nPersentase kelas training:")
print((y_train.value_counts(normalize=True) * 100).round(4))

print("\nPersentase kelas testing:")
print((y_test.value_counts(normalize=True) * 100).round(4))
```

```python
# ==========================================
# PROJECT 1: FRAUD DETECTION
# STEP 4 - MODEL TRAINING & EVALUATION
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

# 1. Definisikan model
models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            class_weight="balanced",
            max_iter=2000,
            random_state=42
        ))
    ]),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        class_weight="balanced_subsample",
        random_state=42,
        n_jobs=-1
    )
}

results = []
trained_models = {}

# 2. Training dan evaluasi
for name, model in models.items():
    print(f"\n{'=' * 50}")
    print(f"MODEL: {name}")
    print("=" * 50)

    # Training hanya dengan data training
    model.fit(X_train, y_train)

    # Prediksi data testing
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    # Hitung metrik
    precision = precision_score(
        y_test, y_pred, zero_division=0
    )
    recall = recall_score(
        y_test, y_pred, zero_division=0
    )
    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )
    ap = average_precision_score(y_test, y_prob)

    results.append({
        "Model": name,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "PR-AUC (AP)": ap
    })

    trained_models[name] = model

    print(classification_report(
        y_test,
        y_pred,
        target_names=["Normal", "Fraud"],
        zero_division=0
    ))

    # Confusion matrix
    cm = confusion_matrix(y_test, y_pred)

    ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Normal", "Fraud"]
    ).plot(values_format="d")

    plt.title(f"Confusion Matrix - {name}")
    plt.show()

# 3. Bandingkan performa seluruh model
results_df = pd.DataFrame(results).set_index("Model")

print("\n=== MODEL COMPARISON ===")
display(results_df.round(4))

# 4. Tampilkan model dengan F1-score tertinggi
best_model_name = results_df["F1-Score"].idxmax()

print("Model dengan F1-score tertinggi:", best_model_name)
print(
    "Catatan: model terbaik belum ditentukan hanya dari F1-score; "
    "periksa juga Recall, Precision, dan PR-AUC."
)
```

```python

# ==========================================
# PROJECT 1: FRAUD DETECTION
# STEP 5 - THRESHOLD OPTIMIZATION
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    precision_recall_curve,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    classification_report,
    ConfusionMatrixDisplay
)

# 1. Siapkan fitur dan target
X = df_clean.drop(columns=["Class"])
y = df_clean["Class"]

# 2. Split menjadi 70% train, 15% validation, 15% test
X_train_val, X_test_final, y_train_val, y_test_final = (
    train_test_split(
        X, y, test_size=0.15, random_state=42, stratify=y
    )
)

X_train_final, X_val_final, y_train_final, y_val_final = (
    train_test_split(
        X_train_val,
        y_train_val,
        test_size=0.1764705882,
        random_state=42,
        stratify=y_train_val
    )
)

print("Train:", X_train_final.shape)
print("Validation:", X_val_final.shape)
print("Test:", X_test_final.shape)

# 3. Train Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_final, y_train_final)

# 4. Prediksi probabilitas pada validation set
val_prob = rf_model.predict_proba(X_val_final)[:, 1]

# Evaluasi threshold default 0.5
val_pred_default = (val_prob >= 0.5).astype(int)

# 5. Cari threshold dengan F1-score tertinggi
precision, recall, thresholds = precision_recall_curve(
    y_val_final, val_prob
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-12)
)

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

val_pred_tuned = (val_prob >= best_threshold).astype(int)

print("\n=== VALIDATION RESULTS ===")
print("Threshold default:", 0.5)
print("Threshold optimal:", round(best_threshold, 4))

for label, pred in [
    ("Default 0.5", val_pred_default),
    ("Optimized threshold", val_pred_tuned)
]:
    print(f"\n{label}")
    print("Precision:", round(
        precision_score(y_val_final, pred, zero_division=0), 4
    ))
    print("Recall:", round(
        recall_score(y_val_final, pred, zero_division=0), 4
    ))
    print("F1-score:", round(
        f1_score(y_val_final, pred, zero_division=0), 4
    ))

# 6. Plot hubungan threshold dengan precision dan recall
plt.figure(figsize=(9, 5))
plt.plot(thresholds, precision[:-1], label="Precision")
plt.plot(thresholds, recall[:-1], label="Recall")
plt.plot(thresholds, f1_scores, label="F1-score")
plt.axvline(
    best_threshold,
    linestyle="--",
    label=f"Best threshold = {best_threshold:.3f}"
)
plt.xlabel("Classification Threshold")
plt.ylabel("Score")
plt.title("Threshold Optimization on Validation Set")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# 7. Evaluasi final pada test set satu kali
test_prob = rf_model.predict_proba(X_test_final)[:, 1]
test_pred = (test_prob >= best_threshold).astype(int)

print("\n=== FINAL TEST RESULTS ===")
print("Threshold:", round(best_threshold, 4))
print("PR-AUC (AP):", round(
    average_precision_score(y_test_final, test_prob), 4
))

print("\nClassification Report:")
print(classification_report(
    y_test_final,
    test_pred,
    target_names=["Normal", "Fraud"],
    zero_division=0
))

ConfusionMatrixDisplay.from_predictions(
    y_test_final,
    test_pred,
    display_labels=["Normal", "Fraud"],
    values_format="d"
)
plt.title("Random Forest - Final Test Confusion Matrix")
plt.show()
```

```python

# ==========================================
# PROJECT 1: FRAUD DETECTION
# STEP 6 - SAVE PORTFOLIO ARTIFACTS
# ==========================================

from pathlib import Path
import json
import joblib
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    precision_recall_curve
)

# 1. Buat folder portofolio di Google Drive
portfolio_dir = Path(
    "/content/drive/MyDrive/portfolio/fraud-detection-credit-card"
)
portfolio_dir.mkdir(parents=True, exist_ok=True)

# 2. Hitung metrik final
cm = confusion_matrix(y_test_final, test_pred)
tn, fp, fn, tp = cm.ravel()

test_metrics = {
    "model": "Random Forest",
    "threshold": float(best_threshold),
    "precision_fraud": float(
        precision_score(y_test_final, test_pred, zero_division=0)
    ),
    "recall_fraud": float(
        recall_score(y_test_final, test_pred, zero_division=0)
    ),
    "f1_fraud": float(
        f1_score(y_test_final, test_pred, zero_division=0)
    ),
    "pr_auc_average_precision": float(
        average_precision_score(y_test_final, test_prob)
    ),
    "true_positive": int(tp),
    "false_positive": int(fp),
    "true_negative": int(tn),
    "false_negative": int(fn)
}

# 3. Simpan metrik dalam JSON
with open(portfolio_dir / "final_metrics.json", "w") as f:
    json.dump(test_metrics, f, indent=4)

# 4. Simpan classification report
report = classification_report(
    y_test_final,
    test_pred,
    target_names=["Normal", "Fraud"],
    output_dict=True,
    zero_division=0
)

pd.DataFrame(report).T.to_csv(
    portfolio_dir / "classification_report.csv"
)

# 5. Simpan hasil prediksi test
predictions = pd.DataFrame({
    "transaction_row_id": X_test_final.index,
    "actual_class": y_test_final.to_numpy(),
    "fraud_probability": test_prob,
    "predicted_class": test_pred
})

predictions.to_csv(
    portfolio_dir / "test_predictions.csv",
    index=False
)

# 6. Simpan confusion matrix
fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Normal", "Fraud"]
).plot(ax=ax, values_format="d", colorbar=False)

ax.set_title("Random Forest - Final Test Confusion Matrix")
fig.tight_layout()
fig.savefig(
    portfolio_dir / "confusion_matrix.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# 7. Simpan grafik optimasi threshold pada validation set
precision, recall, thresholds = precision_recall_curve(
    y_val_final, val_prob
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-12)
)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(thresholds, precision[:-1], label="Precision")
ax.plot(thresholds, recall[:-1], label="Recall")
ax.plot(thresholds, f1_scores, label="F1-score")
ax.axvline(
    best_threshold,
    linestyle="--",
    label=f"Selected threshold = {best_threshold:.3f}"
)

ax.set_title("Threshold Optimization - Validation Set")
ax.set_xlabel("Threshold")
ax.set_ylabel("Score")
ax.legend()
ax.grid(alpha=0.3)
fig.tight_layout()
fig.savefig(
    portfolio_dir / "threshold_optimization.png",
    dpi=150,
    bbox_inches="tight"
)
plt.show()
plt.close(fig)

# 8. Buat business insights report
insights = f"""
# Fraud Detection - Business Insights

## Project Overview
This project builds a machine learning model to identify
potentially fraudulent credit card transactions.

## Dataset
- Source: Kaggle Credit Card Fraud Detection
- Original dataset: 284,807 transactions
- Rows removed as exact duplicates: 1,081
- Cleaned dataset: 283,726 transactions
- Fraud cases after cleaning: 473

## Final Model
- Algorithm: Random Forest
- Decision threshold: {best_threshold:.4f}
- Threshold selected using the validation set.
- Final performance evaluated on a separate test set.

## Final Test Results
- Precision (fraud): {test_metrics["precision_fraud"]:.4f}
- Recall (fraud): {test_metrics["recall_fraud"]:.4f}
- F1-score (fraud): {test_metrics["f1_fraud"]:.4f}
- PR-AUC (Average Precision): {test_metrics["pr_auc_average_precision"]:.4f}

## Confusion Matrix Counts
- True positives: {tp}
- False positives: {fp}
- True negatives: {tn}
- False negatives: {fn}

## Business Interpretation
The model identifies {tp} fraudulent transactions in the test set.
It misses {fn} fraudulent transactions and flags {fp} legitimate
transactions for further review.

A lower threshold may increase fraud detection but can also
increase false alarms. The preferred operating threshold should
reflect the relative cost of missed fraud and manual investigations.

## Limitations
- Results are based on a public, anonymized dataset.
- The test set uses a stratified random split, not a chronological split.
- Model performance does not guarantee equivalent performance
  on real-world banking transactions.
- Model predictions should support, not replace, risk investigation.
"""

(portfolio_dir / "business_insights.md").write_text(
    insights.strip() + "\n",
    encoding="utf-8"
)

# 9. Simpan model agar dapat digunakan kembali
joblib.dump(
    rf_model,
    portfolio_dir / "random_forest_model.joblib"
)

# 10. Buat draft README untuk GitHub
readme = f"""
# Credit Card Fraud Detection

A machine learning portfolio project for identifying potentially
fraudulent credit card transactions using Random Forest.

## Business Problem

Fraudulent transactions are rare compared with legitimate transactions.
A useful detection system must evaluate fraud precision and recall,
rather than relying on accuracy alone.

## Dataset

- Source: [Kaggle Credit Card Fraud Detection](https://www.kaggle.com/datasets/mlg-ulb/creditcardfraud)
- Original records: 284,807
- Records after exact-duplicate removal: 283,726
- Fraud cases after cleaning: 473

The raw dataset is not included in this repository. Download it from
Kaggle and follow the notebook's data-loading instructions.

## Methodology

1. Data inspection and exploratory data analysis.
2. Duplicate inspection and data cleaning.
3. Stratified train, validation, and test split.
4. Baseline comparison: Logistic Regression and Random Forest.
5. Random Forest threshold selection using validation data.
6. Final evaluation on the held-out test set.

## Final Test Results

| Metric | Score |
|---|---:|
| Fraud precision | {test_metrics["precision_fraud"]:.4f} |
| Fraud recall | {test_metrics["recall_fraud"]:.4f} |
| Fraud F1-score | {test_metrics["f1_fraud"]:.4f} |
| PR-AUC (Average Precision) | {test_metrics["pr_auc_average_precision"]:.4f} |
| Selected threshold | {best_threshold:.4f} |

## Key Findings

- Fraud detection is an imbalanced classification problem.
- Threshold selection changes the balance between recall and false alarms.
- The final model's threshold was selected on validation data, then
  evaluated on the separate test set.
- The model is a portfolio demonstration, not a production banking system.

## Project Files

- `01_fraud_detection_eda_modeling.ipynb`: main analysis notebook
- `business_insights.md`: business findings and limitations
- `final_metrics.json`: final test metrics
- `classification_report.csv`: classification metrics
- `confusion_matrix.png`: final test confusion matrix
- `threshold_optimization.png`: validation threshold analysis

## Tools

Python, Pandas, NumPy, Matplotlib, Seaborn, Scikit-learn.

## Reproducibility

Use Google Colab or a compatible Python environment. Download the dataset
from the source linked above and update the input path as needed.
"""

(portfolio_dir / "README.md").write_text(
    readme.strip() + "\n",
    encoding="utf-8"
)

# 11. Tampilkan hasil akhir dan daftar file
print("=== FINAL TEST METRICS ===")
for key, value in test_metrics.items():
    print(f"{key}: {value}")

print("\n=== SAVED PORTFOLIO FILES ===")
for file in sorted(portfolio_dir.iterdir()):
    print(f"{file.name} ({file.stat().st_size / 1024:.1f} KB)")

print("\nFolder portofolio:", portfolio_dir)
```
